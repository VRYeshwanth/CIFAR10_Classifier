```python
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
```

```python
labels = pd.read_csv('../cifar-10/trainLabels.csv')
print(labels.head())
print(f"Shape: {labels.shape}")
print(f"Columns: {labels.columns}")
```

```python
print(f"Unique Labels:-\n{labels['label'].unique()}")
print(f"\n\nValue Counts:-\n{labels["label"].value_counts()}")
```

```python
image_id = 1
graphTitle = labels[labels['id'] == image_id]['label'].values[0]
image = Image.open(f'../cifar-10/train/{image_id}.png')
print(image)
print(f"Image size: {image.size}")
print(f"Image Mode: {image.mode}")

plt.imshow(image)
plt.title(graphTitle)
plt.axis("off")
plt.show()
```
